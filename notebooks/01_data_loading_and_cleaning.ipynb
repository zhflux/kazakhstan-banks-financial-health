```python
# Загружаем с правильной строки, колонки 0,1,2 = номер, банк, активы
df_clean = pd.read_excel("../data/raw/nbk_2023_capital_assets.xls", header=None, skiprows=8)

# Оставим только нужные колонки
df_clean = df_clean[[0, 1, 2]].copy()

# Дадим понятные названия
df_clean.columns = ["number", "bank_name", "assets"]

# Уберём пустые строки
df_clean = df_clean.dropna(subset=["bank_name"])

df_clean.head(10)
```

```python
import glob

all_files = {
    2023: "../data/raw/nbk_2023_capital_assets.xls",
    2024: "../data/raw/nbk_2024_capital_assets.xls",
    2025: "../data/raw/nbk_2025_capital_assets.xls",
    2026: "../data/raw/nbk_2026_capital_assets.xls",
}

frames = []

for year, path in all_files.items():
    df_temp = pd.read_excel(path, header=None, skiprows=8)
    df_temp = df_temp[[0, 1, 2]].copy()
    df_temp.columns = ["number", "bank_name", "assets"]
    df_temp = df_temp.dropna(subset=["bank_name"])
    df_temp["year"] = year
    frames.append(df_temp)

df_all = pd.concat(frames, ignore_index=True)
print(df_all.shape)
df_all.head(10)
```

```python
# Сохраняем в processed
df_all.to_csv("../data/processed/banks_financial_health_clean.csv", index=False)
print("Сохранено!")
```

```python
frames = []

for year, path in all_files.items():
    df_temp = pd.read_excel(path, header=None, skiprows=8)
    df_temp = df_temp[[0, 1, 2]].copy()
    df_temp.columns = ["number", "bank_name", "assets"]
    
    # Убираем строки где number не число
    df_temp = df_temp[pd.to_numeric(df_temp["number"], errors="coerce").notna()]
    
    # Убираем строки где assets пустой
    df_temp = df_temp[df_temp["assets"].notna()]
    
    df_temp["year"] = year
    frames.append(df_temp)

df_all = pd.concat(frames, ignore_index=True)
print(df_all.shape)
df_all
```
